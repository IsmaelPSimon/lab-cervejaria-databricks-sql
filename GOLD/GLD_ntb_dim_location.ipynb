```python
df = spark.sql("""
    select distinct
        md5(country || '-' || state || '-' || city) AS sk_localidade,
        country as nom_pais,
        state as nom_estado,
        city as nom_cidade
    from
        silver.breweries               
""")
```

```python
# df.printSchema()
```

```python
df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("gold.dim_localidades")
```
