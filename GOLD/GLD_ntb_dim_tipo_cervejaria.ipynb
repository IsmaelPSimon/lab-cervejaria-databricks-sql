```python
df = spark.sql("""
    select distinct
        md5(brewery_type) as sk_tipo_cervejaria,
        brewery_type as des_tipo_cervejaria,
        CASE brewery_type
            WHEN 'Closed' THEN 'Fechada'
            WHEN 'Bar' THEN 'Bar'
            WHEN 'Micro' THEN 'Microcervejaria'
            WHEN 'Regional' THEN 'Cervejaria regional'
            WHEN 'Large' THEN 'Cervejaria de grande porte'
            WHEN 'Nano' THEN 'Nanocervejaria'
            WHEN 'Proprietor' THEN 'Proprietária individual'
            WHEN 'Planning' THEN 'Em planejamento'
            WHEN 'Brewpub' THEN 'Brewpub'
            WHEN 'Contract' THEN 'Cervejaria contratada'
            WHEN 'Location' THEN 'Unidade'
            WHEN 'Taproom' THEN 'Taproom'
            WHEN 'Beergarden' THEN 'Jardim da cerveja'
            ELSE 'Outro'
        END AS des_tipo_cervejaria_traduzido
    from
        silver.breweries;
""")
```

```python
df.printSchema()
```

```python
df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("gold.dim_tipo_cervejaria")
```
