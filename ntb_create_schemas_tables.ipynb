```python
%sql
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
```

```python
%sql
drop TABLE if exists silver.breweries;
drop TABLE if exists gold.dim_localidades;
drop TABLE if exists gold.dim_tipo_cervejaria;
drop TABLE if exists gold.dim_cervejarias;
```

```python
%sql
CREATE TABLE silver.breweries (
  id string comment "Identificador unico",
  name string comment "Nome da cervejaria",
  brewery_type string comment "Tipo de cervejaria",
  address_1 string comment "Endereço part 1",
  address_2 string comment "Endereço part 2",
  address_3 string comment "Endereço part 3",
  city string comment "Cidade",
  state_province string comment "Provincia",
  postal_code string comment "Código postal",
  country string comment "País de origem",
  longitude double comment "Longitude",
  latitude double comment "Latitude",
  phone string comment "Telefone",
  website_url string comment "Site da cervejaria",
  state string comment "Estado",
  street string comment "Endereço Completo ou Complemento"
) USING DELTA
PARTITIONED BY (country);
```

```python
%sql
create table gold.dim_localidades(
  sk_localidade string comment "Identificador Unico",
  nom_pais string comment "Nome do País",
  nom_estado string comment "Nome de Estado",
  nom_cidade string comment "Nome da Cidade",
  CONSTRAINT pk_localidade PRIMARY KEY (sk_localidade)
)
using delta;
```

```python
%sql
create table if not exists gold.dim_tipo_cervejaria(
  sk_tipo_cervejaria string comment "Identificador Unico",
  des_tipo_cervejaria string comment "Tipo de cervejaria",
  des_tipo_cervejaria_traduzido string comment "Tradução do tipo de cervejaria",
  CONSTRAINT pk_tipo_cervejaria PRIMARY KEY (sk_tipo_cervejaria)
)
using delta;
```

```python
%sql
CREATE TABLE if not exists gold.dim_cervejarias (
  sk_cervejaria STRING,
  nom_cervejaria STRING,
  sk_tipo_cervejaria STRING,
  sk_localidade STRING,
  longitude DOUBLE,
  latitude DOUBLE,
  cod_postal STRING,
  des_endereco_1 STRING,
  des_endereco_2 STRING,
  des_endereco_3 STRING,
  nro_telefone STRING,
  url_site STRING,
  CONSTRAINT pk_cervejaria PRIMARY KEY (sk_cervejaria),
  CONSTRAINT fk_tipo_cervejaria FOREIGN KEY (sk_tipo_cervejaria) REFERENCES gold.dim_tipo_cervejaria (sk_tipo_cervejaria),
  CONSTRAINT fk_localidade FOREIGN KEY (sk_localidade) REFERENCES gold.dim_localidades (sk_localidade)
)
USING DELTA;
```
