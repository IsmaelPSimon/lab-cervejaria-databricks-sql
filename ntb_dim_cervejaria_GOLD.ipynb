```python
df = spark.sql("""
select
  a.id as sk_cervejaria, 
  a.name as nom_cervejaria,
  b.sk_tipo_cervejaria,
  c.sk_localidade,
  a.longitude,
  a.latitude,
  a.postal_code as cod_postal,
  a.address_1 as des_endereco_1,
  a.address_2 as des_endereco_2,
  a.address_3 as des_endereco_3,
  a.phone as nro_telefone,
  a.website_url as url_site
from
  silver.breweries a
  inner join gold.dim_tipo_cervejaria b on (b.des_tipo_cervejaria = a.brewery_type)
  inner join gold.dim_localidades c on (c.nom_cidade=a.city and c.nom_estado=a.state and c.nom_pais=a.country)
""")
```

```python
df.printSchema()
```

```python
df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("gold.dim_cervejarias")
```
