```python
df = spark.sql("""
  select
    trim(id) as id,
    trim(name) as name,
    initcap(trim(brewery_type)) as brewery_type,
    trim(address_1) as address_1,
    trim(address_2) as address_2,
    trim(address_3) as address_3,
    trim(city) as city,
    trim(state_province) as state_province,
    trim(postal_code) as postal_code,
    trim(country) as country,
    longitude,
    latitude,
    trim(phone) as phone,
    trim(lower(website_url)) as website_url,
    trim(state) as state,
    trim(street) as street
  from
    bronze.raw_breweries
""")
```

```python
# display(df)
```

```python
df.write \
.mode("overwrite") \
.format("delta") \
.partitionBy("country") \
.saveAsTable("silver.breweries")
```
